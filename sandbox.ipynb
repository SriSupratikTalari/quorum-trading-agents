```python
import re
import requests
from bs4 import BeautifulSoup

url='https://finance.yahoo.com/markets/article/apple-microsoft-hit-all-time-highs-as-nasdaq-surges-2-in-post-election-rally-alphacheck-152548814.html'
r=requests.get(url,headers={"User-Agent":"Mozilla/5.0"},timeout=15)
r.raise_for_status()
soup=BeautifulSoup(r.text,"html.parser")
body=soup.find("div",class_="body-wrap yf-jo6sif") or soup
paras=[re.sub(r"\s+"," ",p.get_text()).strip() for p in body.find_all("p")]
print("\n\n".join(p for p in paras if p))
```
